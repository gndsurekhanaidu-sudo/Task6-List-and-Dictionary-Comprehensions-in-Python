```python
squares = []

for i in range(1, 6):
    squares.append(i * i)

print(squares)
```

```python
squares = [i * i for i in range(1, 6)]
print(squares)
```

```python
numbers = [1, 2, 3, 4, 5]

squares = [x * x for x in numbers]

print("Squares:", squares)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even = [x for x in numbers if x % 2 == 0]

print("Even numbers:", even)
```

```python
words = ["python", "java", "data"]

upper_words = [word.upper() for word in words]

print(upper_words)
```

```python
numbers = [2, 4, 6, 8, 10]

result = [x for x in numbers if x > 5]

print(result)
```

```python
numbers = [1, 2, 3, 4, 5]

result = [x + 10 for x in numbers]

print(result)
```

```python
numbers = [1, 2, 3, 4, 5]

squares = {x: x*x for x in numbers}

print(squares)
```

```python
marks = {"Ravi": 85, "Sita": 92, "Ram": 70}

result = {name: mark for name, mark in marks.items() if mark >= 80}

print(result)
```

```python
names = ["ravi", "sita", "ram"]

result = {name: name.upper() for name in names}

print(result)
```
